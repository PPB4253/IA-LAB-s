Vamos a implementar un algoritmo genético y a utilizarlo para obtener el valor que maximiza diferentes funciones. En todos los casos, los genes que utilizaremos serán los valores 0 y 1, la longitud de los cromosomas será 10 y la función decodifica será la función binario_a_decimal presentada a continuación:

```python
def binario_a_decimal(l):
    return sum(x*(2**i) for (i,x) in enumerate(l)) 
```

El método fitness dependerá de la función que queramos maximizar. Por ejemplo, para maximizar la siguiente función:

$$f_1(x) = -x^2$$

Utilizaremos la sigiuente función en Python, que recibiendo un cromosoma x nos devolverá el valor de fitness correspondiente:

```python
def f1(x):
    return -binario_a_decimal(x)**2
```

Otro ejemplo, para maximizar la siguiente función:
$$f_2(x) = -(x - 2)^2 + 5$$

Utilizaremos la sigiuente función en Python, que recibiendo un cromosoma x nos devolverá el fitness correspondiente:

```python
def f2(x):
    return -(binario_a_decimal(x)-2)**2+5
```

Otro ejemplo, para maximizar la siguiente función:
    $$ f_3(x) = -x^2 + 4x $$
    
Utilizaremos la sigiuente función en Python, que recibiendo un cromosoma x nos devolverá el fitness correspondiente:

```python
def f3(x):
    return -(binario_a_decimal(x)-2)**2+4*binario_a_decimal(x)
```
