## Parte I: Problema del viajante como problema de búsqueda local


El problema del viajante se describe como sigue:

*Tienes un conjunto de 𝑛 ciudades y las distancias entre cada par de ellas. El objetivo es encontrar una ruta que visite todas las ciudades exactamente una vez y regrese a la ciudad de origen, minimizando la distancia total recorrida. Modela este problema para ser resuelto mediante el algoritmo de escalada con reinicio, definiendo la representación de la solución, la función de valoración y el esquema de vecindad.*

En nuestro caso, una solución vendrá dada por una permutación de las ciudades, que indicará el orden en que se recorren (desde la útima ciudad volvemos a la primera).

### Ejercicio 1 [0.25 ptos]

Implementar el método *genera_estado_inicial(ciudades)* que dada una lista de ciudades construye, aleatoriamente, una permutación de las ciudades.

```python
# Solución:

def genera_estado_inicial(ciudades):
    estado_inicial = list(ciudades)
    random.shuffle(estado_inicial)
    return estado_inicial
```

Ejemplos de uso:

```python
genera_estado_inicial(andalucia.keys())
#Salida (el contenido de la lista puede variar, es aleatorio):
#['sevilla',  'cordoba',  'almeria',  'huelva',  'cadiz',  'malaga',  'jaen',  'granada']
```

```python
genera_estado_inicial(["a","b","c","d"])
#Salida (el contenido de la lista puede variar, es aleatorio):
#['d', 'c', 'a', 'b']
```

### Ejercicio 2 [0.25 ptos]

Implementar el método *genera_sucesor_intercambio(estado)* que construya un circuito a partir de uno dado, intercambiando la posición de dos ciudades elegidas aleatoriamente.  

**Nota:** El estado devuelto por este método debe ser **diferente** al estado original. 

```python
# Solución:

def genera_sucesor_intercambio(estado):
    nuevo_estado = list(estado)
    ultimo_indice = len(nuevo_estado) - 1

    i = random.randint(0, ultimo_indice)        # Elige valor aleatorio entre 0 y ultimo_indice
    j = random.randint(0, ultimo_indice)        # Elige valor aleatorio entre 0 y ultimo_indice

    while (i == j):
        j = random.randint(0, ultimo_indice)    # Elige valor aleatorio entre 0 y ultimo_indice

    nuevo_estado[i], nuevo_estado[j] = nuevo_estado[j], nuevo_estado[i]

    return nuevo_estado
```

Ejemplos de uso:

### Ejercicio 3 [0.75 ptos]

Implementar el método *genera_sucesor_bloque(estado)* que construya un circuito a partir de uno dado, extrayendo un bloque de ciudades consecutivas de tamaño aleatorio y reubicándolo al final del circuito.

**Nota:** El estado devuelto por este método debe ser **diferente** al estado original. 

```python
# Solución:

def genera_sucesor_bloque(estado):
    estado_original = list(estado)
    nuevo_estado = list(estado)

    while (nuevo_estado == estado_original):

        numero_aleatorio1 = random.randint(0, len(estado_original))
        numero_aleatorio2 = random.randint(0, len(estado_original))

        if (numero_aleatorio1 > numero_aleatorio2):
            grande = numero_aleatorio1
            chico = numero_aleatorio2
        else:
            grande = numero_aleatorio2
            chico = numero_aleatorio1

        primer_cacho = estado_original[:chico]
        segundo_cacho = estado_original[grande:]
        bloque = estado_original[chico:grande]

        nuevo_estado = primer_cacho + segundo_cacho + bloque

    return nuevo_estado
```

```python
genera_sucesor_bloque(['sevilla',  'cordoba',  'almeria',  'huelva',  'cadiz',  'malaga',  'jaen',  'granada'])
#Salida (puede variar, el bloque que se coloca al final de la lista es aleatorio):
#['sevilla',  'granada',  'cordoba',  'almeria',  'huelva',  'cadiz',  'malaga',  'jaen']
```

### Ejercicio 4 [0.75 ptos]

Implementar el método *valoracion(estado,diccionario_ciudades)*, este método debe devolver la longitud total del recorrido representado por *estado* utilizando el diccionario de ciudades pasado como segundo parámetro. Dicho diccionario tiene como claves los nombres de las ciudades del circuito y sus valores son su coordenadas.

```python
# Solución:

def valoracion(estado,diccionario_ciudades):
    resultado = 0

    for i in range(len(estado) - 1):
        ciudad1 = estado[i]
        ciudad2 = estado[i + 1]

        coord_ciudad1 = diccionario_ciudades.get(ciudad1)
        coord_ciudad2 = diccionario_ciudades.get(ciudad2)

        x1, y1 = coord_ciudad1
        x2, y2 = coord_ciudad2

        distancia = math.sqrt((x2-x1)**2 + (y2-y1)**2)      # DISTANCIA ENTRE 2 CIUDADES

        resultado += distancia

    ultima_ciudad = estado[-1]      #PQ ES UN CIRCUITO CERRADO
    primera_ciudad = estado[0]

    coord_ultima = diccionario_ciudades.get(ultima_ciudad)
    coord_primera = diccionario_ciudades.get(primera_ciudad)

    xult, yult = coord_ultima
    xpri, ypri = coord_primera

    distancia_cierre = math.sqrt((xpri - xult)**2 + (ypri - yult)**2)      # DISTANCIA ENTRE 2 CIUDADES

    resultado += distancia_cierre


    return resultado
```

## Parte II: Implementación del algoritmo de escalada con reinicio aleatorio para el problema del viajante

### Ejercicio 5 [0.75 ptos]

Definir la función *escalada(diccionario_ciudades,funcion_sucesor)* que haciendo uso de los métodos implementados anteriormente implemente el método de búsqueda en escalada para tratar de buscar una solución al problema del viajante representado a través de *diccionario_ciudades* utilizando la función *funcion_sucesor* como método para generar sucesores. El diccionario tiene como claves los nombres de las ciudades del circuito y sus valores son su coordenadas y la función *funcion_sucesor* será una de las funciones implementadas en el Ejercicio 2 y Ejercicio 3.

Seguir el siguiente esquema:

```python
# -------------------------------------------------------------------
# ESCALADA()
# 1. Hacer ACTUAL igual GENERA-ESTADO-INICIAL() y 
# VALOR-ACTUAL igual a F-VALORACIÓN(ACTUAL)
# 2. Hacer VECINO igual a GENERA-SUCESOR(ACTUAL) y
# VALOR-VECINO igual a F-VALORACIÓN(VECINO)
# 3. Mientras VALOR-VECINO sea MENOR O IGUAL que VALOR-ACTUAL,
#    3.1 Hacer ACTUAL igual a VECINO
#    y VALOR-ACTUAL igual a VALOR-VECINO.
#    3.2 Hacer VECINO igual a GENERA-SUCESOR(ACTUAL) y
#    VALOR-VECINO igual a F-VALORACIÓN(VECINO)
# 4. Devolver ACTUAL y VALOR-ACTUAL
```

```python
# Solución:

def escalada(diccionario_ciudades,funcion_sucesor):
    # 1.
    actual = genera_estado_inicial(diccionario_ciudades.keys())
    valor_actual = valoracion(actual, diccionario_ciudades)

    # 2.
    vecino = funcion_sucesor(actual)
    valor_vecino = valoracion(vecino, diccionario_ciudades)

    # 3.
    while (valor_vecino <= valor_actual):
        # 3.1
        actual = vecino
        valor_actual = valor_vecino
        # 3.2
        vecino = funcion_sucesor(actual)
        valor_vecino = valoracion(vecino, diccionario_ciudades)

    # 4.
    return (actual, valor_actual)
```

### Ejercicio 6 [0.5 ptos]

Definir la función *escalada_con_reinicio(k,diccionario_ciudades,funcion_sucesor)* que haciendo uso de los métodos implementados anteriormente implemente el método de búsqueda en escalada con reinicio aleatorio (haciendo *k* reinicios) para tratar de buscar una solución al problema del viajantes representado a través de *diccionario_ciudades* haciendo uso de la fnción *funcion_sucesor* para generar sucesores. El método debe devolver la mejor solución encontrada y la valoración correspondiente.

```python
# Solución:

def escalada_con_reinicio(k,diccionario_ciudades,funcion_sucesor):
    valor = 0
    mejor_valor = 999999
    mejor_actual = None

    while valor < k:
        actual, valor_actual = escalada(diccionario_ciudades, funcion_sucesor)

        if (valor_actual < mejor_valor) :
            mejor_valor = valor_actual
            mejor_actual = actual

        valor = valor + 1

    return (mejor_actual, mejor_valor)
```

### Ejercicio 7 [0.25 ptos]

Implementa una función llamada `comparador(diccionario_ciudades,k,num_exp)` que ejecute *num_exp* veces el algoritmo de escalada con reinicio (con *k* reinicios) utilizando tanto el método *genera_sucesor_intercambio* como el método *genera_sucesor_bloque*. La función debe promediar la longitud de los circuitos obtenidos con cada uno de los métodos genera_sucesor y devolver dichos promedios.

```python
# Solución:

def comparador(diccionario_ciudades,k,num_exp):
    valor = 0
    suma1 = 0
    suma2 = 0
    promedio = 0

    while valor < num_exp:
        ciudades1, distancia1 = escalada_con_reinicio(k, diccionario_ciudades, genera_sucesor_intercambio)
        ciudades2, distancia2 = escalada_con_reinicio(k,diccionario_ciudades,genera_sucesor_bloque)

        suma1 += distancia1
        suma2 += distancia2

        valor = valor + 1

    promedio1 = suma1 / num_exp
    promedio2 = suma2 / num_exp

    return (promedio1, promedio2)
```
