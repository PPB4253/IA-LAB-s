Vamos a usar una estructura de datos prefijada, para representar redes bayesianas. Por ejemplo, la siguiente variable red_alarma contiene la representación de la red bayesiana del ejemplo de la alarma visto en clase

En general, una red bayesiana se representará mediante una tupla de tres elementos. Cada uno de estos elementos representará:

* Las variables aleatorias y sus posibles valores: un diccionario que asocia cada nombre de variable con una lista de sus posible valores. 
* Los padres de cada variable en la red: un diccionario que asocia cada nombre de variable con la lista de sus padres. 
* Las tablas de probabilidad de cada nodo: un diccionario que asocia cada nombre de variable con su tabla de probabilidad. 

A su vez, la tabla de rpobabilidad asociada a cada variable X es un diccionario que asocia cada combinación de valores de los padres con la distribución de probabilidad de X dada esa combinación de valores. 

Por ejemplo, si en la tabla de "Alarma" hay una correspondencia (True,False):[0.94,0.06], quiere decir que: P(Alarma=True|Robo=True,Terremoto=False)=0.94, y que P(Alarma=False|Robo=True,Terremoto=False)=0.06.

### Ejercicio 1

Definir una función "orden_compatible(red)", que recibiendo una red bayesiana, devuelva una ordenación topológica del grafo de la red. Es decir, una secuencia (una lista) de todas las variables de la red, cumpliendo que para toda variable X que sea padre de otra variable Y en la red, X va antes que Y en la secuencia.

```python
# Solución:

def orden_compatible(red):
    res = []

    valores = red[0]
    padres = red[1]
    probabilidad = red[2]

    lista_variables = list(valores.keys())

    while len(res) < len(valores):
        for valor in lista_variables:
            if (all(padre in res for padre in padres[valor])):      # Comprueba si todos los padres de la variable ya están procesados en 'res' (devuelve True automático si es un nodo raíz sin padres). ESTO LO QUE HACE ES AÑADIR VARIABLES SOLO SI ESTAN SUS PADRES YA EN LA LISTA
                res.append(valor)                   # Meto la variable
                lista_variables.remove(valor)       # La elimino de la lista de pendientes
                break

    return res
```

### Ejercicio 2

Definir una función "muestreo(valores,distr)" que recibiendo como entrada una lista de n valores de una variable aleatoria y una distribución de probabilidad (una lista de n probabilidades), devuelve aleatoriamente uno de los valores de la lista. La probabilidad de generar cada uno de los valores debe ser la correspondiente probabilidad en la distribución. 

```python
# Solución:

def muestreo(valores,distr):
    valor_aleatorio = random.random()
    acumulado = 0.0

    for i in range(len(valores)):
        acumulado += distr[i]
        if valor_aleatorio <= acumulado:    # Lo que hacemos del muestro: 0,13 < P(a), pues eligo a a
            return valores[i]

    return valores[-1]  # Retorno de seguridad por si hay imprecisiones de redondeo decimal
```

Ejemplos (téngase en cuenta que son valores aleatorios, por lo que los resultados no siempre son éstos):

```python
muestreo(["v1","v2","v3"],[0.1,0.8,0.1])
# Salida: 'v2'
```

```python
muestreo(["v1","v2","v3"],[0.1,0.8,0.1])
# Salida 'v1'
```

Una vez definida la función, comprobar con un número grande de muestreos que efectivamente son sucesos aleatorios con la probabilidad pretendida. Para comprobar el comportamiento de muestreo, podríamos hacer algo así (téngase en cuenta que los resultados son aleatorios):

```python
sum(muestreo(["v1","v2","v3"],[0.1,0.8,0.1])=="v1" for _ in range(1000000))/1000000
# Salida (aprox.): 0.099943
```

### Ejercicio 3

Definir la función "muestreo_a_priori(red)", que obtiene un evento aleatorio completo, siguiendo la distribución que define una red bayesiana que recibe como entrada (transparencias del tema de teoría)

```python
# Solución:

def muestreo_a_priori(red):
    res = {}

    valores = red[0]
    padres = red[1]
    probabilidad = red[2]

    for variable in orden_compatible(red):      # Cogemos las variables de la funcion de orden de antes, para ver la variable solo si estan los padres en la lista antes
        valores_padres = tuple(res[padre] for padre in padres[variable])   # VEMOS LOS VALORES QUE TOMAN LOS PADRE
        dist_prob = probabilidad[variable][valores_padres]                 # VEMOS LAS PROBABILIDADES (POSITIVAS Y NEGATIVAS)
        res[variable] = muestreo(valores[variable], dist_prob)             # AÑADIMOS AL DICCIONARIO LA VARIABLE CON SU VALOR

    return res
    
```

Ejemplos (téngase en cuenta que son resultados aleatorios):

```python
muestreo_a_priori(red_alarma)
# Salida:
# {'Terremoto': False,'Robo': False, 'Alarma': False, 'Maria': False,'Juan': False}
```

```python
muestreo_a_priori(red_aspersor)
# Salida:
# {'nublado': False, 'lluvia': False, 'aspersor': True, 'hierba mojada': True}
```

```python
muestreo_a_priori(red_infarto)
# Salida:
# {'fumador': True, 'alimentación_equilibrada': True, 'deportista': False, 'hipertenso': True, 'infarto': True}
```

La siguiente función "normaliza(d)" normaliza una distribución de probabilidades, que viene dada mediante un diccionario d:

```python
def normaliza(d):
    suma=sum(d.values())
    return {var:d[var]/suma for var in d}
```

Ejemplos:

```python
normaliza({"v1":0.21,"v2":0.11,"v3":0.37})
# Salida:
# {'v1': 0.30434782608695654,
#  'v2': 0.15942028985507248,
#  'v3': 0.5362318840579711}
```

```python
normaliza({True:0.012,False:0.008})
# Salida:
# {True: 0.6, False: 0.4}
```

### Ejercicio 4

Definir la función "muestreo_con_rechazo(var,observado,red,N)", que hace un cálculo aproximado de la probabilidad P(var|observado), respecto de una red bayesiana dada, usando el método de muestreo con rechazo (ver transparencias del tema 3). El número N indica el número de muestreos que ha de realizar.

```python
# Solución:

def muestreo_con_rechazo(var,observado,red,N):
    res = {val:0 for val in red[0][var]}    # red[0] ACCEDE A LA PRIMERA PARTE DE MI RED BAYESIANA Y [var] ACCEDE A LA VARIABLE ESPECIFICA var Y EXTRAE TODOS LOS VALORES QUE PUEDE TOMAR
            # val: 0 inicializa los valores que puede tomar la variable a 0, Ej: Si llamas a la función preguntando por la variable "Robo", esta línea genera de golpe el diccionario: res = {True: 0, False: 0}
    numero = 0

    while numero < N:
        muestra = muestreo_a_priori(red)
        numero = numero + 1
        if all(muestra[k] == v for k, v in observado.items()):  # ESTO HACE QUE LA MUESTRA QUE ACABO DE GENERAR SEA RECHAZADA O NO, MIRA LOS VALORES DE LAS VARIABLES DE EVIDENCIA (observado), Y COMPARA SI LOS VALORES ALEATORIOS GENERADOS EN MI MUESTRA COINCIDEN
            res[muestra[var]] += 1  # AÑADIMOS 1 AL LADO DE LA VARIABLE QUE SEA, SI ES POSITIVA AL TRUE Y SI ES NEGATIVA AL FALSE

    # NORMALIZAMOS
    if sum(res.values()) > 0:
        return normaliza(res)
    else:
        return res
```

### Ejercicio 5

El principal problema del algoritmo de muestreo con rechazo es que se pueden generar muchas muestras que se rechazan, cuando éstas no sean compatibles con los valores de las variables observadas.

Comprobar esto experimentalmente. Para ello, modificar la función muestreo_con_rechazo del ejercicio anterior, para que además de calcular la probabilidad aproximada, escriba por pantalla el total de muestras que se han rechazado.

```python
# Solución:

def muestreo_con_rechazo_bis(var,observado,red,N):
    res = {val:0 for val in red[0][var]}    # red[0] ACCEDE A LA PRIMERA PARTE DE MI RED BAYESIANA Y [var] ACCEDE A LA VARIABLE ESPECIFICA var Y EXTRAE TODOS LOS VALORES QUE PUEDE TOMAR
            # val: 0 inicializa los valores que puede tomar la variable a 0, Ej: Si llamas a la función preguntando por la variable "Robo", esta línea genera de golpe el diccionario: res = {True: 0, False: 0}
    numero = 0
    muestras_rechazadas = 0

    while numero < N:
        muestra = muestreo_a_priori(red)
        numero = numero + 1
        if all(muestra[k] == v for k, v in observado.items()):  # ESTO HACE QUE LA MUESTRA QUE ACABO DE GENERAR SEA RECHAZADA O NO, MIRA LOS VALORES DE LAS VARIABLES DE EVIDENCIA (observado), Y COMPARA SI LOS VALORES ALEATORIOS GENERADOS EN MI MUESTRA COINCIDEN
            res[muestra[var]] += 1  # AÑADIMOS 1 AL LADO DE LA VARIABLE QUE SEA, SI ES POSITIVA AL TRUE Y SI ES NEGATIVA AL FALSE
        else:
            muestras_rechazadas = muestras_rechazadas + 1

    print(f"Rechazadas {muestras_rechazadas} de un total de {N}")   # EL f PARA PONER METER VARIABLES EN EL TEXTO

    # NORMALIZAMOS
    if sum(res.values()) > 0:
        return normaliza(res)
    else:
        return res
```

Para paliar este problema del muestreo con rechazo, se introduce el algoritmo de ponderación por verosimilitud, tal y como se ha visto en la teoría. Se trata de generar directamente muestras compatibles con las observaciones, forzando a que en las variables observadas se tomen los valores observados. Como consecuencia, cada muestra se genera con un "peso", que equivale a la probabilidad de que hubiera ocurrido aleatoriamente aquello que se ha forzado.

### Ejercicio 6

Definir una función "ponderacion_por_verosimilitud(var,observado,red,N)" que implemente un cálculo aproximado de la probabilidad P(var|observado), respecto de una red bayesiana dada, usando el método de ponderación por verosimilitud. El número N indica el número de muestreos que ha de realizar.

```python
# Solución

def ponderacion_por_verosimilitud(var, observado, red, N):
    # Inicializamos el acumulador de pesos a 0.0 para cada valor de la variable de consulta
    res = {val: 0.0 for val in red[0][var]}
    numero = 0
    
    valores = red[0]
    padres = red[1]
    probabilidades = red[2]
    
    # 1. Necesitamos el orden topológico para evaluar padres antes que hijos
    orden = orden_compatible(red)
    
    while numero < N:
        w = 1.0  # El peso inicial de cada muestra es 1
        muestra = {}
        numero = numero + 1
        
        # 2. Generamos la muestra variable por variable siguiendo el orden
        for v in orden:
            # Construimos la tupla con los valores que han tomado los padres en esta muestra
            valores_padres = tuple(muestra[p] for p in padres[v])
            distr = probabilidades[v][valores_padres]
            
            if v in observado:
                # Si es una variable de evidencia: fijamos el valor y multiplicamos el peso
                valor_fijo = observado[v]
                muestra[v] = valor_fijo
                
                # Buscamos qué probabilidad tenía ese valor específico y actualizamos el peso w
                indice_valor = valores[v].index(valor_fijo)
                w = w * distr[indice_valor]
            else:
                # Si es una variable oculta/consulta: la muestreamos aleatoriamente
                muestra[v] = muestreo(valores[v], distr)
                
        # 3. Sumamos el peso final de la muestra al resultado correspondiente
        res[muestra[var]] += w
        
    # 4. Normalizamos los pesos acumulados para obtener la probabilidad final
    if sum(res.values()) > 0:
        return normaliza(res)
    else:
        return res
```
